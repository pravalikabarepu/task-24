```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv(r"C:\Users\Pravalika.b\OneDrive\Documents\store.csv", encoding="latin1")
data.head()
```

```python
print("Rows:",data.shape[0])
print("Columns:",data.shape[1])
```

```python
data.columns=data.columns.str.strip()
```

```python
print("\nColumn names:")
print(data.columns.tolist())
```

```python
print("Number of rows:",len(data))
print("number of columns:",len(data.columns))
```

```python
print("\n data types:")
print(data.dtypes)
```

```python
print("\n missing values:")
print(data.isnull().sum())
```

```python
print("\n duplicate rows:", data.duplicated().sum())
```

```python
if "Order Date" in data.columns:
    data["Order Date"]=pd.to_datetime(
        data["Order Date"],
        errors="coerce"
    )
if "Ship Date" in data.columns:
    data["Ship Date"]=pd.to_datetime(
        data["Ship Date"],
        errors="coerce"
    )
```

```python
audit_results=[]
```

```python
missing_count=data.isnull().sum().sum()
audit_results.append({
    "Rule":"Missing Values",
    "Description":"check for missing values in the dataset",
    "Issue_Count":missing_count,
    "Status":"PASS" if missing_count==0 else "FAIL"
})
```

```python
duplicate_count=data.duplicated().sum()
audit_results.append({
    "Rule":"Duplicate Rows",
    "Description":"check for completely duplicated records",
    "Issue_Count":duplicate_count,
    "Status":"PASS" if duplicate_count==0 else "FAIL"
})
```

```python
if "Sales" in data.columns:
    negative_sales=(data["Sales"]<0).sum()
else:
    negative_sales=0
audit_results.append({
    "Rule":"Negative Values",
    "Description":"Sales should not be negative",
    "Issue_Count":negative_sales,
    "Status":"PASS" if negative_sales==0 else "FAIL"
})
```

```python
if "Quantity" in data.columns:
    invalid_quantity=(data["Quantity"]<=0).sum()
else:
    invalid_quantity=0
audit_results.append({
    "Rule":"Invalid Quantity",
    "Description":"Quantity sholud be greater than zero",
    "Issue_Count":invalid_quantity,
    "Status":"PASS" if invalid_quantity==0 else "FAIL"
})
```

```python
if "Discount" in data.columns:
    invalid_discount=(
        (data["Discount"]<0) |
        (data["Discount"]>1)
    ).sum()
else:
    invalid_discount=0
audit_results.append({
    "Rule":"Invalid Discount",
    "Description":"Discount sholud be between 0 and 1",
    "Issue_Count":invalid_discount,
    "Status":"PASS" if invalid_discount==0 else "FAIL"
  })   
    
```

```python
if "profit" in data.columns:
    missing_profit=data["Profit"].isnull().sum()
else:
    missing_profit=0
audit_results.append({
    "Rule":"Missing Profit",
    "Description":"Profit should not be missing",
    "Issue_Count":missing_profit,
    "Status":"PASS" if missing_profit==0 else "FAIL"
}) 
```

```python
if "Order Date" in data.columns:
    invalid_order_date=data["Order Date"].isnull().sum()
else:
    invalid_order_date=0
audit_results.append({
    "Rule":"Invalid Order Date",
    "Description":"Order date should contain valid dates",
    "Issue_Count":invalid_order_date,
    "Status":"PASS" if invalid_order_date==0 else "FAIL"
})
```

```python
if "Ship Date" in data.columns:
    invalid_ship_date=data["Ship Date"].isnull().sum()
else:
    invalid_ship_date=0
audit_results.append({
    "Rule":"Invalid Ship Date",
    "Description":"Ship date sholud contain valid dates",
    "Issue_Count":invalid_ship_date,
    "Status":"PASS" if invalid_ship_date==0 else "FAIL"
})
```

```python
if "Oder Date" in data.columns and "Ship Date" in data.columns:
    data_consistency=(
        data["Ship Date"] < data["Order Date"]
    ).sum()
else:
    data_consistency=0
audit_results.append({
    "Rule":"Data Consistency",
    "Description":"Ship date should not be earlier than order date",
    "Issue_Count":data_consistency,
    "Status":"PASS" if data_consistency==0 else "FAIL"
})
```

```python
if "Customer ID" in data.columns:
    missing_customer_id=data["Customer ID"].isnull().sum()
else:
    missing_customer_id=0
audit_results.append({
    "Rule":"Missing Customer ID",
    "Description":"Customer ID should not be missing",
    "Issue_Count":missing_customer_id,
    "Status":"PASS" if missing_customer_id==0 else "FAIL"
})
```

```python
audit_report=pd.DataFrame(audit_results)
```

```python
print(audit_report)
```

```python
issue_log=audit_report[
    audit_report["Issue_Count"]>0
].copy()
```

```python
print("\n ISSUE LOG")
```

```python
if len(issue_log)==0:
    print("no data quality issues found.")
else:
    print(issue_log)
```

```python
missing_report=pd.DataFrame({
    "Column": data.columns,
    "missing_count":data.isnull().sum().values
})
```

```python
missing_report=missing_report[
    missing_report["missing_count"]>0
].sort_values(
    by="missing_count",
    ascending=False
)
print("\n missing value report")
```

```python
duplicate_rows=data[data.duplicated(keep=False)].copy()
print("\n DUPLICATE REPORT")
```

```python
print("Duplicate rows:", len(duplicate_rows))
```

```python
if len(duplicate_rows)>0:
    print(duplicate_rows.head(10))
```

```python
range_report=[]
if "Sales" in data.columns:
    range_report.append({
        "Field":"Sales",
        "Minimum":data["Sales"].min(),
        "Maximum":data["Sales"].max(),
        "Invalid_Count": (data["Sales"]<0).sum()
    })
```

```python
if "Qunantity" in data.columns:
    range_report.append({
        "Field":"Qunatity",
        "Minimum":data["Quantity"].min(),
        "Maximum":data["Qunatity"].max(),
         "Invalid_Count": (data["Qunatity"]<=0).sum()
     })

   
```

```python
if "Discount" in data.columns:
    range_report.append({
        "Field":"Discount",
        "Minimum":data["Discount"].min(),
        "Maximum":data["Discount"].max(),
         "Invalid_Count": (data["Discount"] <0) |
         (data["Discount"]>1)
     })
```

```python
range_report=pd.DataFrame(range_report)
print("\n RANGE REPORT")
print(range_report)
```

```python
cleaned_data=data.copy()
```

```python
cleaned_data=cleaned_data.drop_duplicates()
```

```python
if "Sales" in cleaned_data.columns:
    cleaned_data=cleaned_data[
        cleaned_data["Sales"]>=0
    ]
```

```python
if "Quantity" in cleaned_data.columns:
    cleaned_data=cleaned_data[
        cleaned_data["Quantity"]>0
    ]
```

```python
if "Discount" in cleaned_data.columns:
    cleaned_data=cleaned_data[
        (cleaned_data["Discount"]>=0) &
        (cleaned_data["Discount"]<=1)
    ]
```

```python
if "Order Date" in cleaned_data.columns and "Ship date" in cleaned_data.columns:
    cleaned_data=cleaned_data[
        (
            cleaned_data["Ship Date"].isnull()
        )|
        (
            cleaned_data["Order Date"].isnull()
        )|
        (
            cleaned_data["Ship Date"]>=cleaned_data["Order Date"]
        )
        
        
    ]
```

```python
print("\n CLEANED DATA")
```

```python
print("original rows:",len(data))
print("Cleaned rows:",len(cleaned_data))
print("Rows removed:",len(data)-len(cleaned_data))
```

```python
cleaned_sample=cleaned_data.head(100)
print("\n CLEANED SAMPLE")
print(cleaned_sample.head(10))
```

```python
print("\n FINAL QUALITY CHECK")
```
